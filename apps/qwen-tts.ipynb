```python
import torch
import soundfile as sf
from qwen_tts import Qwen3TTSModel
```

```python
model_path = r"E:/models/Qwen3-TTS-12Hz-1.7B-CustomVoice"

model = Qwen3TTSModel.from_pretrained(
    model_path,
    device_map="cuda:0",
    dtype=torch.bfloat16,
    # attn_implementation="flash_attention_2",
    attn_implementation="sdpa",
)
```

```python
for _speaker_ in model.get_supported_speakers():
    # single inference
    wavs, sr = model.generate_custom_voice(
        text="Using Qwen TTS model for translating english to tamil if tamil is supported",
        language="English", # Pass `Auto` (or omit) for auto language adaptive; if the target language is known, set it explicitly.
        speaker=_speaker_,
        instruct="Talk in soft voice", # Omit if not needed.
    )
    sf.write(f"{_speaker_} - english soft output_custom_voice.mp3", wavs[0], sr)
```
